```python

import pandas as pd
import os

base_dir = "/content/drive/MyDrive/archive (1)"
csv_dir = os.path.join(base_dir, "csv")

mass_train = pd.read_csv(
    os.path.join(csv_dir, "mass_case_description_train_set.csv")
)

mass_test = pd.read_csv(
    os.path.join(csv_dir, "mass_case_description_test_set.csv")
)

calc_train = pd.read_csv(
    os.path.join(csv_dir, "calc_case_description_train_set.csv")
)

calc_test = pd.read_csv(
    os.path.join(csv_dir, "calc_case_description_test_set.csv")
)

print("All four CSV files loaded successfully!")

print("Mass training:", mass_train.shape)
print("Mass testing:", mass_test.shape)
print("Calcification training:", calc_train.shape)
print("Calcification testing:", calc_test.shape)
```

```python

datasets = {
    "Mass Training": mass_train,
    "Mass Testing": mass_test,
    "Calcification Training": calc_train,
    "Calcification Testing": calc_test
}

for name, df in datasets.items():

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())

    print("\nPathology distribution:")
    print(df["pathology"].value_counts(dropna=False))
```

```python

# Inspect unique values in important columns

feature_columns = [
    "breast_density",
    "left or right breast",
    "image view",
    "abnormality type",
    "mass shape",
    "mass margins",
    "calc type",
    "calc distribution",
    "pathology"
]

for name, df in datasets.items():

    print("\n" + "=" * 55)
    print(name)
    print("=" * 55)

    for col in feature_columns:
        if col in df.columns:
            print(f"\n{col}:")
            print(df[col].value_counts(dropna=False))
```

```python

mass_train_patients = set(mass_train["patient_id"])
mass_test_patients = set(mass_test["patient_id"])

calc_train_patients = set(calc_train["patient_id"])
calc_test_patients = set(calc_test["patient_id"])

print(
    "Mass train-test patient overlap:",
    len(mass_train_patients & mass_test_patients)
)

print(
    "Calcification train-test patient overlap:",
    len(calc_train_patients & calc_test_patients)
)

train_patients = mass_train_patients | calc_train_patients
test_patients = mass_test_patients | calc_test_patients

print(
    "Combined train-test patient overlap:",
    len(train_patients & test_patients)
)
```

```python

# Check cross-category patient overlap

mass_train_ids = set(mass_train["patient_id"])
mass_test_ids = set(mass_test["patient_id"])

calc_train_ids = set(calc_train["patient_id"])
calc_test_ids = set(calc_test["patient_id"])

print("Mass training vs Calcification testing:")
print(mass_train_ids & calc_test_ids)

print("\nCalcification training vs Mass testing:")
print(calc_train_ids & mass_test_ids)
```

```python

# Add abnormality type to identify each record

mass_train["record_type"] = "mass"
mass_test["record_type"] = "mass"

calc_train["record_type"] = "calcification"
calc_test["record_type"] = "calcification"

# Combine all four datasets
combined_df = pd.concat(
    [mass_train, mass_test, calc_train, calc_test],
    ignore_index=True,
    sort=False
)

# Standardize pathology labels
combined_df["pathology"] = (
    combined_df["pathology"].str.strip().str.upper()
)

# Convert pathology into binary labels
label_mapping = {
    "BENIGN": 0,
    "BENIGN_WITHOUT_CALLBACK": 0,
    "MALIGNANT": 1
}

combined_df["target"] = combined_df["pathology"].map(
    label_mapping
)

print("Combined dataset shape:", combined_df.shape)

print("\nAbnormality distribution:")
print(combined_df["record_type"].value_counts())

print("\nBinary target distribution:")
print(combined_df["target"].value_counts(dropna=False))

print("\nUnique patients:", combined_df["patient_id"].nunique())
```

```python

from sklearn.model_selection import train_test_split

# Check for missing labels or patient IDs
print("Missing patient IDs:", combined_df["patient_id"].isna().sum())
print("Missing target labels:", combined_df["target"].isna().sum())

# Create one row per patient for stratification
patient_summary = (
    combined_df.groupby("patient_id")["target"]
    .max()
    .reset_index()
)

# Split patients: 70% training, 30% remaining
train_patients, remaining_patients = train_test_split(
    patient_summary,
    test_size=0.30,
    random_state=42,
    stratify=patient_summary["target"]
)

# Split remaining patients equally into validation and testing
val_patients, test_patients = train_test_split(
    remaining_patients,
    test_size=0.50,
    random_state=42,
    stratify=remaining_patients["target"]
)

# Get patient IDs
train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

# Assign complete records to each partition
train_df = combined_df[
    combined_df["patient_id"].isin(train_ids)
].copy()

val_df = combined_df[
    combined_df["patient_id"].isin(val_ids)
].copy()

test_df = combined_df[
    combined_df["patient_id"].isin(test_ids)
].copy()

# Display results
print("\nTraining:", train_df.shape)
print("Validation:", val_df.shape)
print("Testing:", test_df.shape)

print("\nUnique patients:")
print("Training:", train_df["patient_id"].nunique())
print("Validation:", val_df["patient_id"].nunique())
print("Testing:", test_df["patient_id"].nunique())

# Verify no patient overlap
print("\nPatient overlap checks:")
print("Train-Val:", len(train_ids & val_ids))
print("Train-Test:", len(train_ids & test_ids))
print("Val-Test:", len(val_ids & test_ids))

print("\nTraining label distribution:")
print(train_df["target"].value_counts())

print("\nValidation label distribution:")
print(val_df["target"].value_counts())

print("\nTesting label distribution:")
print(test_df["target"].value_counts())
```

```python
metadata_df = metadata_df.copy()

# Mass-specific features
mass_features = ["mass shape", "mass margins"]

for col in mass_features:
    metadata_df[col] = metadata_df[col].fillna(
        metadata_df["abnormality type"].map({
            "mass": "UNKNOWN",
            "calcification": "NOT_APPLICABLE"
        })
    )

# Calcification-specific features
calc_features = ["calc type", "calc distribution"]

for col in calc_features:
    metadata_df[col] = metadata_df[col].fillna(
        metadata_df["abnormality type"].map({
            "mass": "NOT_APPLICABLE",
            "calcification": "UNKNOWN"
        })
    )

print("Missing values after handling:\n")
print(metadata_df.isnull().sum())
```

```python
# Separate metadata using the existing patient-level split

train_meta = metadata_df[
    metadata_df["patient_id"].isin(train_ids)
].copy()

val_meta = metadata_df[
    metadata_df["patient_id"].isin(val_ids)
].copy()

test_meta = metadata_df[
    metadata_df["patient_id"].isin(test_ids)
].copy()

print("Train:", train_meta.shape)
print("Validation:", val_meta.shape)
print("Test:", test_meta.shape)
```

```python
print("Train-Val overlap:",
      len(set(train_meta["patient_id"]) & set(val_meta["patient_id"])))

print("Train-Test overlap:",
      len(set(train_meta["patient_id"]) & set(test_meta["patient_id"])))

print("Val-Test overlap:",
      len(set(val_meta["patient_id"]) & set(test_meta["patient_id"])))
```

```python
# Input features
feature_cols = [
    "breast_density_final",
    "left or right breast",
    "image view",
    "abnormality type",
    "mass shape",
    "mass margins",
    "calc type",
    "calc distribution"
]

# Separate X and y
X_train = train_meta[feature_cols]
y_train = train_meta["target"]

X_val = val_meta[feature_cols]
y_val = val_meta["target"]

X_test = test_meta[feature_cols]
y_test = test_meta["target"]

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Separate numerical and categorical columns
numeric_features = ["breast_density_final"]

categorical_features = [
    "left or right breast",
    "image view",
    "abnormality type",
    "mass shape",
    "mass margins",
    "calc type",
    "calc distribution"
]

# Numerical preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform validation and test data
X_val_processed = preprocessor.transform(X_val)
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed validation shape:", X_val_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from PIL import Image, ImageOps
from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
import torch
import numpy as np

IMAGE_SIZE = 224

image_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Image transformations created successfully!")
```

```python

from torch.utils.data import Dataset
import torch

class CBISMammogramDataset(Dataset):

    def __init__(self, dataframe, metadata_features,
                 transform=None):

        self.df = dataframe.reset_index(drop=True)
        self.metadata = np.asarray(
            metadata_features, dtype=np.float32
        )

        self.transform = transform

        assert len(self.df) == len(self.metadata)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):

        row = self.df.iloc[index]

        image_path = row["actual_path"]

        # Load mammogram image
        with Image.open(image_path) as img:

            img = ImageOps.exif_transpose(img)

            # Convert grayscale to RGB
            img = img.convert("RGB")

            # Resize while preserving aspect ratio.
            # Remaining area is padded with black pixels.
            img = ImageOps.pad(
                img,
                (IMAGE_SIZE, IMAGE_SIZE),
                method=Image.Resampling.LANCZOS,
                color=(0, 0, 0)
            )

            if self.transform:
                img = self.transform(img)

        # Corresponding metadata
        metadata = torch.tensor(
            self.metadata[index],
            dtype=torch.float32
        )

        # Binary target
        label = torch.tensor(
            float(row["target"]),
            dtype=torch.float32
        )

        case_id = str(row["case_id"])

        return img, metadata, label, case_id


print("Custom mammogram dataset class created!")
```

```python

# Convert sparse matrices to dense arrays if necessary

def to_dense_array(data):
    if sparse.issparse(data):
        return data.toarray().astype(np.float32)

    return np.asarray(data, dtype=np.float32)


X_train_array = to_dense_array(X_train_processed)
X_val_array = to_dense_array(X_val_processed)
X_test_array = to_dense_array(X_test_processed)

print("Metadata arrays:")
print(X_train_array.shape)
print(X_val_array.shape)
print(X_test_array.shape)
```

```python

train_dataset = CBISMammogramDataset(
    image_train_df,
    X_train_array,
    transform=image_transform
)

val_dataset = CBISMammogramDataset(
    image_val_df,
    X_val_array,
    transform=image_transform
)

test_dataset = CBISMammogramDataset(
    image_test_df,
    X_test_array,
    transform=image_transform
)

print("All three datasets created successfully!")

print("Training:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Testing:", len(test_dataset))
```

```python

def verify_dataloader(name, loader):

    total_images = 0

    for images, metadata, labels, case_ids in loader:

        # Check image dimensions
        assert images.shape[1:] == (3, 224, 224)

        # Check metadata dimensions
        assert metadata.shape[1] == 101

        # Check for invalid numerical values
        assert torch.isfinite(images).all()
        assert torch.isfinite(metadata).all()
        assert torch.isfinite(labels).all()

        # Check labels
        assert torch.all((labels == 0) | (labels == 1))

        total_images += len(labels)

    print(f"{name} verification successful!")
    print("Images processed:", total_images)


verify_dataloader("Training", train_loader)
verify_dataloader("Validation", val_loader)
verify_dataloader("Testing", test_loader)
```
